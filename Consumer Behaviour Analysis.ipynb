```python
def age_group(age): #feature engineering customer age group
    if age <= 25:
        return 'Gen Z'
    elif age <= 35:
        return 'Young Adult'
    elif age <= 45:
        return 'Mid Career'
    elif age <= 55:
        return 'Mature Adults'
    else:
        return 'Seniors'

customers['Age_Group'] = customers['Age'].apply(age_group)
customers['Age_Group'].value_counts().sort_values(ascending=False)
```

```python
def year_customer_range(years): #feature engineering years_customer_range
    if years == 0:
        return 'New Customer'
    elif years <= 2:
        return '1-2 Years'
    elif years <=5:
        return '3-5 Years'
    elif years <=10:
        return '6-10 Years'
    else:
        return '10+ Years'

customers['Years_As_Customer_Range'] = customers['YearsAsCustomer'].apply(year_customer_range)
customers['Years_As_Customer_Range'].value_counts().sort_values(ascending=False)
```

```python
products['Discount_Percentage'] = ( #creating new column Discount %
    (products['OriginalPrice'] - products['CurrentPrice'])
    / products['OriginalPrice']) * 100

products['Discount_Percentage'] = products['Discount_Percentage'].round(2)
products['Discount_Percentage']
```

```python
def decision_time_range(time): #feature engineering a decision_time_range column
    if time < 10:
        return 'Fast'
    elif time < 25:
        return 'Medium'
    else:
        return 'Slow'

transactions['Decision_Speed'] = transactions['DecisionTimeMinutes'].apply(decision_time_range)

transactions['Decision_Speed'].value_counts().sort_values(ascending=False)
```

```python
merged_data = pd.merge(transactions, customers, on='CustomerID', how='left') #Merging all three datasets
merged_data = pd.merge(merged_data, products, on='ProductID', how='left')

merged_data.head()
```
